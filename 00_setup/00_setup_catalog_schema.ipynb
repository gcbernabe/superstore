## 1. Create Catalog

```python
spark.sql(f"CREATE CATALOG IF NOT EXISTS {CATALOG}")
spark.sql(f"USE CATALOG {CATALOG}")
print(f"Catalog '{CATALOG}' is ready")
```

## 2. Create Schemas (Medallion Layers)

```python
for schema in [SCHEMA_BRONZE, SCHEMA_SILVER, SCHEMA_GOLD]:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{schema}")
    print(f"Schema '{CATALOG}.{schema}' is ready")
```

## 3. Create Volume (for raw file uploads)

```python
spark.sql(f"""
    CREATE VOLUME IF NOT EXISTS {CATALOG}.{SCHEMA_BRONZE}.raw_data
""")
print(f"✅ Volume '{CATALOG}.{SCHEMA_BRONZE}.raw_data' is ready.")
print(f"   Upload your raw files to: {VOLUME_PATH}")
```
