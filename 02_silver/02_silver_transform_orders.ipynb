### 1. Read from Bronze

```python
log("Reading from Bronze ...")
df_bronze = spark.table(TBL_BRONZE_RAW)
```

```python
display(df_bronze.limit(10))
```

### 2. Extract Order Columns & Add Ingestion Metadata

```python
df_orders = df_bronze.select(
    F.col("row_id"),
    F.col("order_id"),
    F.col("order_date"),
    F.col("ship_date"),
    F.col("ship_mode"),
    F.col("customer_id"),
    F.col("product_id"),
    F.col("country"),
    F.col("city"),
    F.col("state"),
    F.col("postal_code"),
    F.col("region"),
    F.col("sales"),
    F.col("quantity"),
    F.col("discount"),
    F.col("profit"),
    F.col("ingested_at"),
    F.col("file_path"),
    F.col("file_name"),
    F.col("file_size")
) \
.withColumn("transformed_at", F.current_timestamp()) 
display(df_orders.limit(10))
```

```python
df_orders.printSchema()
```

### 3. Change Data Type 

```python
df_orders = df_orders \
    .withColumn("order_date", F.try_to_date(F.col("order_date"), "M/d/yyyy")) \
    .withColumn("ship_date", F.try_to_date(F.col("ship_date"), "M/d/yyyy")) \
    .withColumn("sales", F.col("sales").cast("double")) \
    .withColumn("quantity", F.col("quantity").cast("int")) \
    .withColumn("discount", F.col("discount").cast("double")) \
    .withColumn("profit", F.col("profit").cast("double"))

display(df_orders.limit(10))
df_orders.printSchema()
```

```python
df_orders.display()
```

## 4. Deduplicate and Remove Null Value

```python
log(f"Rows before cleanup: {df_orders.count():,}")

df_silver_orders = (
    df_orders
    .dropna(subset=["order_id", "product_id"])
    .dropDuplicates(["order_id", "product_id"])
)

log(f"Rows after cleanup: {df_silver_orders.count():,}")
```

## 5. Write to Silver

```python
log(f"Writing to {TBL_SILVER_ORDERS} [mode=overwrite] ...")
df_silver_orders.write \
    .format("delta") \
    .option("mergeSchema", "true") \
    .mode("overwrite") \
    .saveAsTable(TBL_SILVER_ORDERS)

count = get_row_count(TBL_SILVER_ORDERS)
log(f"✅ Done. {count:,} rows written to {TBL_SILVER_ORDERS}")
```
